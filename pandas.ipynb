```python
import pandas as pd

# Create a sample DataFrame

data = {
    'Name': ['Sansa', 'Jon', 'Cersei', 'Ned Stark'],
    'Age': [25, 30, 35, 40],
    'City': ['Winterfell', 'Dragonstone', 'King\'s Landing', 'Winterfell']
}

print(type(data))
df = pd.DataFrame(data)
```

```python
df[['Name']]
```

```python
# selecting 2 columns
df[['Name', 'City']] # use double brackets to select multiple columns
```

```python
df[df['Age'] > 30]
```

```python
# filtering rows based on a condition
df[df['Age'] > 30] # Select rows where Age is greater than 30
```

```python
# Get only selected columns with a condition
df[df['Age'] > 30][['Name', 'City']] # Select Name and City for rows where Age is greater than 30
```
