```python
import pandas as pd

df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
df.head()
```

```python
pd.set_option('display.max_columns', None)
df.head()
```

```python
# common operations with pandas
df['Attrition'].value_counts() # count the number of occurrences of each value in the 'Attrition' column
```

```python
df.dtypes # check the data types of each column
```

```python
# update Attrition column to be 1 for Yes and 0 for No
df['Attrition_1'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
df[['Attrition', 'Attrition_1']].head()
```

```python
# or you can use apply with a lambda function
df['Attrition_2'] = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
df[['Attrition', 'Attrition_1', 'Attrition_2']].head()
```

```python
# or you can use np.where
import numpy as np
df['Attrition_3'] = np.where(df['Attrition'] == 'Yes', 1, 0)
```

```python
# or you can also use .loc
df.loc[df['Attrition'] == 'Yes', 'Attrition_4'] = 1
df.loc[df['Attrition'] == 'No', 'Attrition_4'] = 0
```

```python
df.columns
```

```python
cols = [col for col in df.columns if 'Attrition' in col]
df[cols].head()
```

```python
# get rid of all columns we created.
cols_to_drop = [col for col in df.columns if 'Attrition_' in col]
# .drop drops the columns.
# inplace=True means we want to modify the original dataframe
#  instead of creating a new one.
df.drop(columns=cols_to_drop, inplace=True)
```

```python
df.columns
```

```python
# if column does not exist, it will raise an error.
df.drop(columns=['Attrition_1'], inplace=True)
```

```python
df['Attrition'].dtype
```

```python
# modify Attrition
df.loc[df['Attrition'] == 'Yes', 'Attrition'] = '1'
df.loc[df['Attrition'] == 'No', 'Attrition'] = '0'

df['Attrition'] = df['Attrition'].astype(int)
```

```python
# check the distribution of Attrition
df.groupby('Attrition').size()
```

```python
# which Department has the highest attrition rate?
df.groupby('Department').agg({'Attrition': 'sum'}).sort_values(by='Attrition', ascending=False)
```
