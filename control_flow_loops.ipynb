````python
# range function is used to generate a sequence of numbers.
range(5) # generates numbers from 0 to 4
````

````python
# if we want to start from a specific number

for i in range(4, 10):
    print(i)
````

# pass, break, continue

```
break - stops the loop
pass - does nothing, just a place holder
continue - skips the current iteration step, does not execute t
```

````python
for i in range(5):
    if i == 3:
        break
    print(i) # this part is not executed after the for loop is broken.
    print(i)
    print(4+5)


# notice there are no prints after 3 because we broke out of the loop
# when i reached 3
````

````python
for i in range(5):
    if i == 3:
        continue # skips the rest of the block of code in iteration.
    print(i)

# notice iteration 3 is skipped because we used continue statement
# to skip the rest of the code in that iteration 
# and move to the next iteration
````

````python
for i in range(5):
    if i == 3:
        c = 100
        pass # continues to execute the rest of the block of code in iteration.
    print(i)
````
