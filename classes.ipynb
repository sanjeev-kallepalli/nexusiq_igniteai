```python
# Class is a blueprint and object is an instance of a class.

class Car:
    # initialize a class
    def __init__(self, make, model, year):
        self.make = make # attribute
        self.model = model # attribute
        self.year = year # attribute
    
    def display_info(self):
        print(f"Mfg: {self.year} Make: {self.make} Model: {self.model}")
```

```python
# object

my_first_car = Car("Toyota", "Corolla", 2020)
```

```python
my_first_car.make
```

```python
my_first_car.model
```

```python
my_first_car.display_info()
```

```python
my_latest_car = Car("Tesla", "Model 3", 2022)
my_latest_car.display_info()
```

```python
class Student:
    def __init__(self, name, age, admission_no, section, class_name):
        self.name = name
        self.age = age
        self.admission_no = admission_no
        self.section = section
        self.class_name = class_name
```

```python
sanjeev = Student("Sanjeev", 30, "12345", "A", "10th Grade")
```

```python
sanjeev.name
```

```python
# inheritance, Marks class inherits from Student class
class Marks(Student):
    def __init__(self, name, age, admission_no, section, class_name, **marks):
        super().__init__(name, age, admission_no, section, class_name)
        for subject, mark in marks.items():
            setattr(self, subject, mark)
    def get_average(self):
        total_marks = sum(getattr(self, subject) for subject in self.__dict__ if subject not in ['name', 'age', 'admission_no', 'section', 'class_name'])
        num_subjects = len([subject for subject in self.__dict__ if subject not in ['name', 'age', 'admission_no', 'section', 'class_name']])
        return total_marks / num_subjects if num_subjects > 0 else 0
```

```python
sanjeev = Marks("Sanjeev", 20, "A123", "A", "10th Grade", Math=95, Science=90, English=88)
```

```python
sanjeev.English
```

```python
sanjeev.Math
```

```python
sanjeev.get_average()
```

```python
# polymorphism

class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"Hi, I'm {self.name}."
```

```python
# student inherits from person
class Student(Person):   # Student IS-A Person
    def __init__(self, name, grade):
        super().__init__(name)   # call Person.__init__
        self.grade = grade

    def introduce(self):
        # override the method to add more detail
        return f"Hi, I'm {self.name} from grade {self.grade}."
```

```python
# teacher inherits from person
class Teacher(Person):   # Teacher IS-A Person
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def introduce(self):
        return f"Hello, I'm {self.name}, and I teach {self.subject}."
```

```python
ed_stark = Teacher("Eddard Stark", "History")
arya_stark = Student("Arya Stark", "5th Grade")
jon_snow = Person("Jon Snow")
```

```python
ed_stark.introduce()
```

```python
arya_stark.introduce()
```

```python
jon_snow.introduce()
```

```python
people = [ed_stark, arya_stark, jon_snow]
for person in people:
    print(person.introduce())
```
