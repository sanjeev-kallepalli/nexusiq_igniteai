```python
student_master = {
    'Alice': {
        'age': 21,
        'major': 'Computer Science',
        'GPA': 3.8
    },
    'Bob': {
        'age': 22,
        'major': 'Mathematics',
        'GPA': 3.6
    },
    'Charlie': {
        'age': 20,
        'major': 'Physics',
        'GPA': 3.9
    },
    'Daniel': {
        'age': 23,
        'major': 'Chemistry',
        'GPA': 3.7
    }
}
```

```python
# Conditions
major_condition = 'Chemistry'
gpa_condition = 3.5

# Check eligibility for Alice
if student_master['Alice']['major'] == major_condition:
    if student_master['Alice']['GPA'] > gpa_condition: # executes if above condition is true
        print("Alice is eligible for the scholarship.") # executes if above condition is true
```

```python
if student_master['Bob']['major'] == major_condition:
    if student_master['Bob']['GPA'] > gpa_condition:
        print("Bob is eligible for the scholarship.")
```

```python
# define the variable that can be replaced with the name of the student
student_name = 'Charlie'
```

```python
if student_master[student_name]['major'] == major_condition:
    if student_master[student_name]['GPA'] > gpa_condition:
        print(f"{student_name} is eligible for the scholarship.")
```

```python
# we use functions when we want to reuse a block of code
# multiple times without having to rewrite it.
def check_scholarship_eligibility(student_name, mjr_cond, gpa_cond):
    """
    Check if a student is eligible for a scholarship based on their major and GPA.
    Parameters:
    student_name (str): The name of the student to check.
    mjr_cond (str): The major condition for eligibility.
    gpa_cond (float): The GPA condition for eligibility.
    Returns:
    None: Prints the eligibility status of the student.
    """
    if student_master[student_name]['major'] == mjr_cond:
        if student_master[student_name]['GPA'] > gpa_cond:
            print(f"{student_name} is eligible for the scholarship.")
        else:
            print(f"{student_name} is not eligible for the scholarship due to GPA.")
    else:
        print(f"{student_name} is not eligible for the scholarship due to major.")
```
