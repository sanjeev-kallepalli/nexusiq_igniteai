```python
# let us take a new dictionary to understand the methods better
# below is a nested dictionary, which is a dictionary that contains another dictionary as its value.
student_master = {
    'Alice': {
        'age': 21,
        'major': 'Computer Science',
        'GPA': 3.8
    },
    'Bob': {
        'age': 22,
        'major': 'Mathematics',
        'GPA': 3.6
    },
    'Charlie': {
        'age': 20,
        'major': 'Physics',
        'GPA': 3.9
    }
}
```

```python
student_master.keys() # get all keys in the dictionary
```

```python
student_master['Alice'] # get the value associated with the key 'Alice', which is another dictionary containing Alice's information.
```

```python
# we could also use get
student_master.get('Alice') # this will return the same result as student_master['Alice'], but it is safer to use get() method because it will return None if the key does not exist, instead of raising a KeyError.
```

```python
# raises error as key 'Sanjeev' does not exist in the dictionary
student_master['Sanjeev']
```

```python
student_age.get('Sanjeev') # this will return None instead of raising an error, because 'Sanjeev' is not a key in the dictionary.
```

```python
# dictionary keys cannot be duplicated, so if we try to add a key that already exists in the dictionary, it will overwrite the existing value associated with that key.
student_master['Alice'] = {
    'age': 21,
    'major': 'Computer Science',
    'GPA': 4.0
}
student_master
```

```python
# case sensitive
student_master['alice'] = {
    'age': 21,
    'major': 'Computer Science',
    'GPA': 4.0
}
student_master
```

#### Set

set is also denoted by {}, except that it does not have key value pairs

```python
fruits = {'apple', 'banana', 'papaya'}
```

```python
# it contains unique items only, so if we try to add a duplicate item to the set, it will not be added to the set.
fruits.add('apple') # use add() method to add an item to the set.
fruits
```

```python
# add multiple items
fruits.update(["grape", "melon"])
fruits
```

```python
# removing an item from the set
# use remove() method to remove an item from the set. 
# This will raise a KeyError if the item is not found in the set.
fruits.remove('banana')
```

```python
fruits.remove('banana')
```

```python
# more safer way is to use discard()
fruits.discard('watermelon') # this will not raise an error even if 'watermelon' is not found in the set.
```

```python
# we could also use pop here.
fruits.pop()
```

```python
# few more set operations

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

# returns a new set that contains all the unique elements from both sets a and b.
a.union(b) 
```

```python
a | b # this is another way to get the union of two sets using the | operator.
```

```python
# we did not assign the result. the original sets a and b remain unchanged.
print(a, '\n', b)
```

```python
# intersection of two sets
a.intersection(b) # returns common elements.
```

```python
# we could also use
a & b # this is another way to get the intersection of two sets using the & operator.
```

```python
# difference
a - b # returns elements that are in a but not in b.
```

```python
b - a # returns elements that are in b but not in a.
```

```python
# get all uncommon elements from both sets a and b
a.symmetric_difference(b) # returns a new set that contains all the elements that are in either set a or set b but not in both.
```

```python
# we could also use `^` operator to get the symmetric difference of two sets.
a ^ b
```

```python
c = a | b
c
```

```python
a.issubset(b) # returns True if set a is a subset of set b, meaning that all elements of set a are also present in set b. Otherwise, it returns False.
```

```python
a.issubset(c)
```

```python
c.issuperset(a)
```

```python
# usage of set to remove duplicates from a list
my_list = [1, 2, 2, 3, 4, 4, 5]

unique_elements = set(my_list) # convert the list to a set to remove duplicates
print(unique_elements)
```

```python
my_list = list(set(my_list)) # convert the list to a set to remove duplicates, back to list
print(my_list)
```
