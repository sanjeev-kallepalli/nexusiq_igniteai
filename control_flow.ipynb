```python
# executes the indented block when the condition satisfies
# as the condition is false, the block will not be executed (no output)
if 10 > 15:
    print("10 is greater than 15") # this line will not be executed as the condition is false
```

```python
# when condition is true, the block will be executed and the output will be printed
if 10 > 5:
    print("10 is greater than 5") # this block is executed as the condition is true
```

```python
# let us take more complex example.
# we will consider student who has major in Chemistry and has GPA greater than 3.5 for scholarship eligibility
student_master = {
    'Alice': {
        'age': 21,
        'major': 'Computer Science',
        'GPA': 3.8
    },
    'Bob': {
        'age': 22,
        'major': 'Mathematics',
        'GPA': 3.6
    },
    'Charlie': {
        'age': 20,
        'major': 'Physics',
        'GPA': 3.9
    },
    'Daniel': {
        'age': 23,
        'major': 'Chemistry',
        'GPA': 3.7
    }
}
```

```python
student_master.keys()
```

```python
student_master['Bob']['age']
```

```python
# Conditions
major_condition = 'Chemistry'
gpa_condition = 3.5

# Check eligibility for Alice
if student_master['Alice']['major'] == major_condition:
    if student_master['Alice']['GPA'] > gpa_condition: # executes if above condition is true
        print("Alice is eligible for the scholarship.") # executes if above condition is true

# Alice is not eligible for the scholarship as her major is not Chemistry, so the above block will not be executed and no output will be printed for Alice.
```

```python
# check eligibility for Daniel
if student_master['Daniel']['major'] == major_condition:
    if student_master['Daniel']['GPA'] > gpa_condition: # executes if above condition is true
        print("Daniel is eligible for the scholarship.") # executes if above condition is true
# Daniel is eligible for the scholarship as his major is Chemistry and his GPA is greater than 3.5, so the above block will be executed and the output will be printed for Daniel.
```

```python
# we can merge both if statements by using logical operator 'and'
if student_master['Daniel']['major'] == major_condition and student_master['Daniel']['GPA'] > gpa_condition:
    print("Daniel is eligible for the scholarship.") # executes if above condition is true
```

```python
if student_master['Alice']['major'] == major_condition and student_master['Alice']['GPA'] > gpa_condition:
    print("Alice is eligible for the scholarship.") # executes if above condition is true
```

```python
# Say we want to relax the condition.
# We want to consider students who have major in Chemistry or have GPA greater than 3.5 for scholarship eligibility
if student_master['Alice']['major'] == major_condition or student_master['Alice']['GPA'] > gpa_condition:
    print("Alice is eligible for the scholarship.") # executes if above condition is true

# one of the condition is true for Alice, so the above block will be executed and the output will be printed for Alice.
```

### if else

```python
# when the if condition is not met, the else is executed if present.

if 10 > 15: # condition is false
    print("10 is greater than 15") # this line will not be executed as the condition is false
else:
    print("10 is not greater than 15") # this line will be executed as the condition is false
```

```python
# only one of the blocks will be executed as the condition is either true or false, not both.
if 10 > 5: # condition is true
    print("10 is greater than 5") # this line will be executed as the condition is true
    print("this is second line in if block")
else:
    print("10 is not greater than 5") # this line will not be executed as the condition is true
    print("this is second line in else block")
```

### if elif else

```python
if a > b:
    print(f"{a} is greater than {b}") # this line will be executed as the condition is true
elif a < b:
    print(f"{a} is less than {b}") # this line will not be executed as the condition is false
else:
    print(f"{a} is equal to {b}") # this line will be executed if neither of the above conditions is true
```

```python
a, b = 10, 10

if a > b:
    print(f"{a} is greater than {b}")
elif a < b:
    print(f"{a} is less than {b}")
else:
    print(f"{a} is equal to {b}")
```

```python
a, b = 1, 10

if a > b:
    print(f"{a} is greater than {b}")
elif a < b:
    print(f"{a} is less than {b}")
else:
    print(f"{a} is equal to {b}")
```

```python
marks = 95

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("D")
```

```python
# note if a condition is satisfied, the other conditions will not be checked.
marks = 99
if marks >= 60:
    print("C")
elif marks >= 75:
    print("B")
elif marks >= 90:
    print("A")
else:
    print("D")
# this is not intended as the conditions are not in the correct order, 
# so the output will be C instead of A, which is incorrect.
```
